```python
#flip sort
def flip(arr,k):
    arr[:k+1]=arr[:k+1][::-1]
def pancakesort(arr):
    n=len(arr)
    for size in range(n,1,-1):
        max_index = arr.index(max(arr[:size]))
        if max_index != size-1:
            if max_index !=0:
                flip(arr,max_index)
            flip(arr,size-1)
n=int(input("enter array size"))
arr=list(map(int,input("enter elements").split()))
pancakesort(arr)
print("Sorted array:",*arr)
```

```python
#4.find the count of occurence of a number using binary search pattern
arr=list(map(int,input("Enter elements:").split()))
left=0
right=len(arr)-1
target=int(input("Enter target:"))
first= -1
while left<=right:
    mid=(left+right)//2
    if arr[mid]==target:
        first=mid
        right=mid-1
    elif arr[mid]<target:
        left=mid+1
    else:
        right=mid-1
print("First occurence index of:",target, "=",first)
#last occurence
left=0
right=len(arr)-1
last= -1
while left<=right:
    mid=(left+right)//2
    if arr[mid]==target:
        last=mid
        left=mid+1
    elif arr[mid]<target:
        left=mid+1
    else:
        right=mid-1
print("last occurence index of:",target, "=",last)
if first==-1:
    print("count:", 0)
else:
    print("count:", last-first+1)
```
