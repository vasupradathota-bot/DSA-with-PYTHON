```python
 # WAP to print the sum of digits
n=int(input("enter a number"))
sum=0
while n!=0:
    d=n%10
    sum+=d
    n//=10
print("sum of digits",sum)
```

```python
#WAP to print reverse 
n=int(input("enter a number"))
rev=0
while n>0:
    d=n%10
    rev = rev *10+d
    n//=10
print("reverse number",rev)
```

```python
n=int(input("enter a number"))
even=0
odd=0
while n!=0:
    d=n%10
    if d%2==0:
        even+=1
    else:
        odd+=1
    n//=10
print("even count",even)
print("odd count",odd)
```

```python
c=input("enter 1st string").replace(' ','').lower()
e=input("enter 2nd string").replace(' ','').lower()
if len(c)!=len(e):
    print("not anagram")
elif sorted(c)==sorted(e):
    print("anagram")
else:
    print("annangramm")

```
