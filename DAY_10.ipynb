```python
#Bubble Sorting
arr = list(map(int,input("Enter elements :").split()))
n=len(arr)
for i in range(n):
    for j in range(0,n-i-1):
        if arr[j]>arr[j+1]:
            arr[j],arr[j+1] =  arr[j+1],arr[j]
print(*arr)
    
```

```python
#selection sorting
arr = list(map(int,input("Enter elements :").split()))
n=len(arr)
for i in range(n):
    minidex=i
    for j in range(i+1,n):
        if arr[j]>arr[minidex]:
            minidex=j
    arr[i],arr[minidex]=  arr[minidex],arr[i]
print(*arr)
    
```
