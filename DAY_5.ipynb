```python
#WAC to find avg of  take n elements in an array and traverse and print them
n=int(input("Enter the number of elements: "))
a=[]
for i in range(n):
    n1=int(input("Enter the value: "))
    a.append(n1)
print("Array Elements : ")
for i in range(n):
    print(a[i],end=' ')
great=a[0]
least=a[0]
for i in range(1,n):
    if a[i]>great:
        great=a[i]
    else:
        a[i]<least
        least=a[i]
print("\nGreatest element is : ",great)
print("Least element is : ",least)
add=0
for i in range(n):
    add+=a[i]
print("Sum of elements is : ",add)
avg=add/n
print("Average is : ",avg)
```

```python
#WAC to read no of even and odd numbers 
n=int(input("Enter the size of the array: "))
arr=[]
for i in range(n):
    num=int(input("Enter a number: "))
    arr.append(num)
num_even=0
num_odd=0
for i in range(n):
    if arr[i]%2==0:
        num_even+=1
    else:
        num_odd+=1
print("The number of even numbers in the array is:", num_even)
print("The number of odd numbers in the array is:", num_odd)
```
