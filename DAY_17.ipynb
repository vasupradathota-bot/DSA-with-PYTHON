```python
# stack operations
# 1.expressions
# 2.parenthesis
# 3.applications
# a+b-> +ab
# infix  a+b*c-
# prefix   +a*bc
# postfix  abc*+

#infix to prefix conversion using stacks
def precedence(op):
    if op=='+'or op=='-':
        return 1
    if op=='*'or op=='/':
        return 2
    if op=='^':
        return 3
    return 0
def infixtoprefix(expression):
    stack=[]
    output=[]
    expression=expression[::-1]
    for ch in expression:
        if ch.isalnum():
            output.append(ch)
        else:
            while stack and precedence(ch)<precedence(stack[-1]):
                output.append(stack.pop())
            stack.append(ch)
    while stack:
        output.append(stack.pop())
    return ''.join(output[::-1])
expression=input("Enter expression:")
prefix=infixtoprefix(expression)
print(prefix)
```
