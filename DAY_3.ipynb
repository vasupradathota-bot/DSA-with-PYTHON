```python
#WAC to print "N" natural numbers using direct recursion
def number(n):
    if n==0:
        return 
    print(n,end =" ")
    number(n-1)
n=int(input("enter a number"))
number(n)
```

```python
#tree recursion with fibonacci number
def fib(n):
    if n<=1:
        return n
    return fib(n-1)+fib(n-2)
n=int(input("enter a number"))
print("fibonacci",fib(n))
```
